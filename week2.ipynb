```python
GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)
MOVES = {
    0: [('Down', 3), ('Right', 1)],
    1: [('Down', 4), ('Left', 0), ('Right', 2)],
    2: [('Down', 5), ('Left', 1)],
    3: [('Up', 0), ('Down', 6), ('Right', 4)],
    4: [('Up', 1), ('Down', 7), ('Left', 3), ('Right', 5)],
    5: [('Up', 2), ('Down', 8), ('Left', 4)],
    6: [('Up', 3), ('Right', 7)],
    7: [('Up', 4), ('Left', 6), ('Right', 8)],
    8: [('Up', 5), ('Left', 7)]
}
def get_neighbors(state):
    blank = state.index(0)
    neighbors = []

    for move, position in MOVES[blank]:
        new_state = list(state)


        new_state[blank], new_state[position] = new_state[position], new_state[blank]
        neighbors.append((tuple(new_state), move))

    return neighbors

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# DFS

def dfs(state, visited, path):
    if state == GOAL:
        return path

    visited.add(state)

    for new_state, move in get_neighbors(state):

        if new_state not in visited:
            result = dfs(new_state, visited, path + [move])
            if result is not None:
                return result

    return None


# IDS


def depth_limited_search(state, depth, path, visited):

    if state == GOAL:
        return path
    if depth == 0:
        return None
    visited.add(state)
    for new_state, move in get_neighbors(state):
        if new_state not in visited:
            result = depth_limited_search(new_state,depth - 1,path + [move],visited)
            if result is not None:
                return result

    return None


def ids(start):
    depth = 0

    while True:
        visited = set()
        result = depth_limited_search(start,depth,[],visited)
        if result is not None:
            return result
        depth += 1
start = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

print("Initial State:")
print_puzzle(start)

# DFS
print("DFS Solution:")

dfs_path = dfs(start, set(), [])

if dfs_path is not None:
    print("Number of moves:", len(dfs_path))
    print("Moves:", dfs_path)
else:
    print("No solution found")


# IDS
print("\nIDS Solution:")

ids_path = ids(start)

if ids_path is not None:
    print("Number of moves:", len(ids_path))
    print("Moves:", ids_path)
else:
    print("No solution found")
```
